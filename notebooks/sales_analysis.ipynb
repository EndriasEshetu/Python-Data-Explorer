## 2. Dataset Description

### 2.1 Data Source

The dataset used in this project is the **Online Retail** dataset from the UCI Machine Learning Repository.

* **Dataset:** Online Retail
* **Source:** UCI Machine Learning Repository
* **Period:** December 2010 – December 2011
* **File format:** Excel (`.xlsx`)
* **Original size:** 541,909 transaction records
* **Number of attributes:** 8

### 2.2 Dataset Overview

The dataset contains transactional records from a UK-based online retail business. Each record represents a product line within an invoice.

### 2.3 Data Dictionary

| Column      | Description                                                                  |
| ----------- | ---------------------------------------------------------------------------- |
| InvoiceNo   | Unique invoice identifier; invoices beginning with C indicate cancellations. |
| StockCode   | Product identifier.                                                          |
| Description | Product name or description.                                                 |
| Quantity    | Number of units purchased or returned.                                       |
| InvoiceDate | Date and time of the transaction.                                            |
| UnitPrice   | Price per unit in GBP (£).                                                   |
| CustomerID  | Customer identifier, where available.                                        |
| Country     | Country associated with the transaction.                                     |

### 2.4 Analytical Scope

The dataset will be used to investigate sales revenue, product performance, monthly sales trends, geographic performance, and transaction data quality.

The dataset does not contain predefined product categories, profit margins, or geographic regions. These will not be assumed in the analysis.
