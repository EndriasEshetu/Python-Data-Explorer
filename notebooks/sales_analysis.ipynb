## 6. Data Cleaning

### 6.1 Introduction

Data cleaning is the process of preparing raw data for reliable analysis. Based on the data-quality investigation, this stage will address duplicate records, missing values, unusual quantities, unusual prices, and cancelled transactions.

The cleaning process will preserve the original dataset and apply documented decisions to a separate working copy.

### 6.2 Cleaning Objectives

* Remove confirmed exact duplicate records.
* Handle records with missing essential information.
* Preserve legitimate returns and cancellations for investigation.
* Identify unusual prices and quantities without making unsupported assumptions.
* Validate the cleaned dataset.
* Save the cleaned data for subsequent analysis.


```python
# Preserve the original dataset
df_clean = df.copy()

# Record the original number of rows
initial_rows = len(df_clean)

print("Original rows:", initial_rows)
print("Working copy created:", len(df_clean))
```

```python
# Count exact duplicates before cleaning
duplicates_before = df_clean.duplicated().sum()

# Remove exact duplicates
df_clean = df_clean.drop_duplicates().copy()

duplicates_removed = (
    initial_rows - len(df_clean)
)

print("Duplicates found:", duplicates_before)
print("Duplicates removed:", duplicates_removed)
print("Rows remaining:", len(df_clean))
```

```python
# Record missing dates
missing_dates_before = df_clean["InvoiceDate"].isna().sum()

# Create a subset suitable for date-based analysis
df_dated = df_clean.dropna(
    subset=["InvoiceDate"]
).copy()

print("Rows with missing dates:", missing_dates_before)
print("Rows with valid dates:", len(df_dated))
```

```python
df_product = df_dated.dropna(
    subset=["Description"]
).copy()

print("Rows available for product analysis:",
      len(df_product))
```

```python
# Identify unusual quantities and prices
df_clean["is_zero_quantity"] = (
    df_clean["Quantity"] == 0
)

df_clean["is_return"] = (
    df_clean["Quantity"] < 0
)

df_clean["is_zero_price"] = (
    df_clean["UnitPrice"] == 0
)

df_clean["is_negative_price"] = (
    df_clean["UnitPrice"] < 0
)

print(
    "Zero quantity:",
    df_clean["is_zero_quantity"].sum()
)

print(
    "Returns or negative quantities:",
    df_clean["is_return"].sum()
)

print(
    "Zero prices:",
    df_clean["is_zero_price"].sum()
)

print(
    "Negative prices:",
    df_clean["is_negative_price"].sum()
)
```

```python
# Identify cancelled invoices
df_clean["is_cancelled"] = (
    df_clean["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

# Create an analysis-specific dataset
sales_clean = df_clean[
    (df_clean["Quantity"] > 0) &
    (df_clean["UnitPrice"] > 0) &
    (~df_clean["is_cancelled"]) &
    (df_clean["InvoiceDate"].notna()) &
    (df_clean["Description"].notna())
].copy()

print("Original rows:", initial_rows)
print("Cleaned sales rows:", len(sales_clean))
```

```python
# Check the cleaned transaction dataset
print("Original rows:", len(df))
print("Cleaned transaction rows:", len(df_clean))
print("Exact duplicates remaining:",
      df_clean.duplicated().sum())

print("Negative quantities retained:",
      (df_clean["Quantity"] < 0).sum())

print("Missing customer IDs retained:",
      df_clean["CustomerID"].isna().sum())

print("Missing dates remaining:",
      df_clean["InvoiceDate"].isna().sum())
```

```python
print("Positive sales rows:", len(sales_clean))

print(
    "Invalid quantities:",
    (sales_clean["Quantity"] <= 0).sum()
)

print(
    "Invalid prices:",
    (sales_clean["UnitPrice"] <= 0).sum()
)

print(
    "Cancelled invoices:",
    sales_clean["is_cancelled"].sum()
)
```

```python
assert not sales_clean.duplicated().any()
assert (sales_clean["Quantity"] > 0).all()
assert (sales_clean["UnitPrice"] > 0).all()
assert (~sales_clean["is_cancelled"]).all()
assert sales_clean["InvoiceDate"].notna().all()
assert sales_clean["Description"].notna().all()

print("All positive-sales validation checks passed!")
```

```python
# Save cleaned transactions, including flags
df_clean.to_csv(
    "../data/cleaned_transactions.csv",
    index=False
)

# Save positive sales for sales analysis
sales_clean.to_csv(
    "../data/cleaned_sales.csv",
    index=False
)

print("Cleaned datasets saved successfully!")
```

## 7. Feature Engineering

Feature engineering is the process of creating new variables from existing data to support analysis.

In this step, we will:

* Calculate transaction revenue.
* Extract year, month, and day from transaction dates.
* Create a year-month feature for monthly sales analysis.
* Prepare separate datasets for sales and returns analysis.

These features will help us identify sales trends, measure revenue, and understand customer purchasing activity.


```python
# Calculate revenue for all cleaned transactions
df_clean["Revenue"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

# Calculate revenue for positive sales
sales_clean["Revenue"] = (
    sales_clean["Quantity"] * sales_clean["UnitPrice"]
)

# Display the results
display(
    sales_clean[
        ["InvoiceNo", "Description", "Quantity",
         "UnitPrice", "Revenue"]
    ].head(10)
)
```

```python
# Ensure InvoiceDate is in datetime format
df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"], errors="coerce"
)

sales_clean["InvoiceDate"] = pd.to_datetime(
    sales_clean["InvoiceDate"], errors="coerce"
)

# Extract date features
for data in [df_clean, sales_clean]:
    data["Year"] = data["InvoiceDate"].dt.year
    data["Month"] = data["InvoiceDate"].dt.month
    data["MonthName"] = data["InvoiceDate"].dt.month_name()
    data["Day"] = data["InvoiceDate"].dt.day
    data["DayOfWeek"] = data["InvoiceDate"].dt.day_name()
    data["YearMonth"] = (
        data["InvoiceDate"].dt.to_period("M")
    )

# Preview engineered features
display(
    sales_clean[
        ["InvoiceDate", "Year", "Month", "MonthName",
         "Day", "DayOfWeek", "YearMonth", "Revenue"]
    ].head(10)
)
```

```python
# Check new columns
feature_columns = [
    "Revenue", "Year", "Month", "MonthName",
    "Day", "DayOfWeek", "YearMonth"
]

display(sales_clean[feature_columns].isnull().sum())

# Check revenue calculation
assert np.allclose(
    sales_clean["Revenue"],
    sales_clean["Quantity"] * sales_clean["UnitPrice"]
)

print("Feature engineering validation completed.")
```

```python
sales_clean.to_csv(
    "../data/engineered_sales.csv",
    index=False
)

print("Engineered sales dataset saved successfully!")
```

### Feature Engineering Summary

In this step, we created:

* **Revenue:** the value of each transaction line.
* **Year:** the year of the transaction.
* **Month:** the numerical month of the transaction.
* **MonthName:** the name of the month.
* **Day:** the day of the month.
* **DayOfWeek:** the weekday of the transaction.
* **YearMonth:** a monthly time period for trend analysis.

These features will be used in the next step to analyze sales performance, product performance, and geographic trends.


## 8. Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is used to summarize and investigate a dataset to discover patterns, trends, and relationships.

In this step, we will analyze:

1. Overall sales performance.
2. Product performance.
3. Monthly sales trends.
4. Sales performance by country.
5. Returns and negative-quantity transactions.

The purpose is to answer business questions using the cleaned and engineered data rather than relying on assumptions.


```python
# Overall sales performance
total_revenue = sales_clean["Revenue"].sum()

total_units_sold = sales_clean["Quantity"].sum()

total_orders = sales_clean["InvoiceNo"].nunique()

average_order_value = (
    sales_clean.groupby("InvoiceNo")["Revenue"].sum().mean()
)

# Net recorded revenue includes negative-quantity transactions
net_recorded_revenue = df_clean["Revenue"].sum()

sales_metrics = pd.DataFrame({
    "Metric": [
        "Gross positive sales revenue",
        "Net recorded revenue",
        "Total units sold",
        "Total orders",
        "Average order value"
    ],
    "Value": [
        total_revenue,
        net_recorded_revenue,
        total_units_sold,
        total_orders,
        average_order_value
    ]
})

display(sales_metrics)
```

```python
# Top 10 products by quantity sold
top_products_quantity = (
    sales_clean.groupby("Description", as_index=False)
    .agg(
        Quantity_Sold=("Quantity", "sum"),
        Revenue=("Revenue", "sum")
    )
    .sort_values("Quantity_Sold", ascending=False)
    .head(10)
)

display(top_products_quantity)
```

```python
# Top 10 products by revenue
top_products_revenue = (
    sales_clean.groupby("Description", as_index=False)
    .agg(
        Revenue=("Revenue", "sum"),
        Quantity_Sold=("Quantity", "sum")
    )
    .sort_values("Revenue", ascending=False)
    .head(10)
)

display(top_products_revenue)
```

```python
# Group revenue and orders by month
monthly_sales = (
    sales_clean.groupby("YearMonth", as_index=False)
    .agg(
        Revenue=("Revenue", "sum"),
        Orders=("InvoiceNo", "nunique"),
        Units_Sold=("Quantity", "sum")
    )
    .sort_values("YearMonth")
)

display(monthly_sales)
```

```python
# Month with the highest revenue
highest_sales_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmax()
]

print("Month with highest revenue:")
display(highest_sales_month)
```

```python
# Month with the lowest revenue
lowest_sales_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmin()
]

print("Month with lowest revenue:")
display(lowest_sales_month)
```

```python
# Sales performance by country
country_sales = (
    sales_clean.groupby("Country", as_index=False)
    .agg(
        Revenue=("Revenue", "sum"),
        Orders=("InvoiceNo", "nunique"),
        Units_Sold=("Quantity", "sum")
    )
    .sort_values("Revenue", ascending=False)
)

# Display the top 10 countries
display(country_sales.head(10))
```

```python
country_sales["Revenue_Percentage"] = (
    country_sales["Revenue"] /
    country_sales["Revenue"].sum() * 100
)

display(country_sales.head(10))
```

```python
# Identify negative-quantity transactions
returns_df = df_clean[
    df_clean["Quantity"] < 0
].copy()

# Summarize negative-quantity transactions
returns_summary = pd.DataFrame({
    "Metric": [
        "Negative-quantity records",
        "Invoices containing negative quantities",
        "Total negative quantity",
        "Recorded value of negative-quantity records"
    ],
    "Value": [
        len(returns_df),
        returns_df["InvoiceNo"].nunique(),
        returns_df["Quantity"].sum(),
        returns_df["Revenue"].sum()
    ]
})

display(returns_summary)
```

```python
# Products with the largest negative recorded values
returns_by_product = (
    returns_df.groupby("Description", as_index=False)
    .agg(
        Negative_Quantity=("Quantity", "sum"),
        Recorded_Value=("Revenue", "sum")
    )
    .sort_values("Recorded_Value")
    .head(10)
)

display(returns_by_product)
```
