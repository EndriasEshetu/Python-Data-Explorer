## 5. Data Quality Investigation

### 5.1 Introduction

Data quality investigation is an important step in preparing data for reliable analysis. The purpose of this stage is to identify missing values, duplicate records, invalid dates, unusual quantities, incorrect prices, and cancelled transactions.

The investigation will help determine which records require cleaning and which unusual values represent legitimate business transactions.

The following areas will be examined:

* Missing values
* Duplicate records
* Unusual quantities
* Invalid or unusual prices
* Missing or invalid dates
* Cancelled transactions

The original dataset will remain unchanged during this investigation.


```python
# Count missing values in each column
missing_count = df.isnull().sum()

# Calculate missing percentages
missing_percentage = (
    df.isnull().mean() * 100
).round(2)

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

display(
    missing_summary.sort_values(
        "Missing Percentage",
        ascending=False
    )
)
```

```python
# Examine records with missing customer IDs
missing_customers = df[df["CustomerID"].isna()]

print("Rows with missing CustomerID:",
      len(missing_customers))

display(missing_customers.head(10))
```

```python
missing_descriptions = df[
    df["Description"].isna()
]

print("Missing descriptions:",
      len(missing_descriptions))

display(missing_descriptions.head(10))
```

```python
# Count exact duplicate rows
duplicate_count = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_count)
```

```python
duplicates = df[df.duplicated(keep=False)]

display(duplicates.head(20))
```

```python
print(
    "Unique invoices:",
    df["InvoiceNo"].nunique()
)

print(
    "Rows with repeated invoice numbers:",
    df["InvoiceNo"].duplicated().sum()
)
```

```python
print("Minimum quantity:", df["Quantity"].min())
print("Maximum quantity:", df["Quantity"].max())

print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())
```

```python
display(
    df[df["Quantity"] < 0]
    [["InvoiceNo", "Description",
      "Quantity", "InvoiceDate", "Country"]]
    .head(20)
)
```

```python
display(
    df.nlargest(10, "Quantity")[
        ["InvoiceNo", "Description",
         "Quantity", "UnitPrice"]
    ]
)
```

```python
print("Minimum price:", df["UnitPrice"].min())
print("Maximum price:", df["UnitPrice"].max())

print("Zero prices:", (df["UnitPrice"] == 0).sum())
print("Negative prices:", (df["UnitPrice"] < 0).sum())
```

```python
display(
    df[df["UnitPrice"] <= 0]
    [["InvoiceNo", "Description",
      "Quantity", "UnitPrice"]]
    .head(20)
)
```

```python
display(
    df.nlargest(10, "UnitPrice")[
        ["InvoiceNo", "Description",
         "Quantity", "UnitPrice"]
    ]
)
```

```python
print("Earliest invoice date:",
      df["InvoiceDate"].min())

print("Latest invoice date:",
      df["InvoiceDate"].max())

print("Missing invoice dates:",
      df["InvoiceDate"].isna().sum())
```

```python
print("InvoiceDate data type:",
      df["InvoiceDate"].dtype)
```

```python
cancelled_mask = (
    df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

cancelled = df[cancelled_mask]

print("Cancelled invoice rows:", len(cancelled))

display(cancelled.head(10))
```

```python
cancellation_summary = pd.DataFrame({
    "Transaction Type": [
        "Cancelled",
        "Not Cancelled"
    ],
    "Rows": [
        cancelled_mask.sum(),
        (~cancelled_mask).sum()
    ]
})

display(cancellation_summary)
```

```python
display(
    df.assign(
        IsCancelled=cancelled_mask,
        QuantityType=np.select(
            [
                df["Quantity"] < 0,
                df["Quantity"] == 0
            ],
            [
                "Negative",
                "Zero"
            ],
            default="Positive"
        )
    ).groupby(
        ["IsCancelled", "QuantityType"]
    ).size().unstack(fill_value=0)
)
```

```python
quality_summary = pd.DataFrame({
    "Issue": [
        "Missing values (all columns)",
        "Exact duplicate rows",
        "Missing CustomerID",
        "Missing Description",
        "Missing InvoiceDate",
        "Negative quantities",
        "Zero quantities",
        "Zero prices",
        "Negative prices",
        "Cancelled invoice rows"
    ],
    "Count": [
        int(df.isnull().sum().sum()),
        int(df.duplicated().sum()),
        int(df["CustomerID"].isna().sum()),
        int(df["Description"].isna().sum()),
        int(df["InvoiceDate"].isna().sum()),
        int((df["Quantity"] < 0).sum()),
        int((df["Quantity"] == 0).sum()),
        int((df["UnitPrice"] == 0).sum()),
        int((df["UnitPrice"] < 0).sum()),
        int(cancelled_mask.sum())
    ]
})

display(quality_summary)
```

### 5.2 Data Quality Findings

The investigation identified several data quality characteristics in the Online Retail dataset.

1. **Missing values:** The investigation identified missing values in CustomerID and Description. The most affected column was CustomerID with 135080 missing records.

2. **Duplicate records:** The dataset contained 5268 exact duplicate rows. Repeated invoice numbers were also observed, but these do not necessarily indicate duplicate transactions.

3. **Unusual quantities:** The dataset contained 10624 negative quantities and 0 zero quantities. Negative quantities require investigation because they may represent product returns.

4. **Unusual prices:** The investigation identified 2515 zero-priced records and 2 negative-priced records. These will be considered separately during cleaning.

5. **Dates:** The dataset covers the period from 2010-12-01 08:26:00 to 2011-12-09 12:50:00. There were 0 missing invoice dates.

6. **Cancellations:** The investigation identified 9288 cancelled invoice rows, which require separate treatment in sales analysis.

These findings will guide the cleaning process. Only records with a clear reason for removal will be excluded, while meaningful business transactions will be retained where possible.
